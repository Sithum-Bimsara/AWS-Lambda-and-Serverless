# 04_Amazon_S3_Bucket

---

## **1. What is Amazon S3?**

Amazon S3 (Simple Storage Service) is an object‑storage service provided by AWS. It allows you to store any type of data (images, videos, documents, backups, logs, datasets) as **objects inside buckets**.

It is highly durable (99.999999999% durability), scalable, and accessible from anywhere.

### **Key Characteristics:**

* **Object Storage:** Data is stored as objects, not files or blocks.
* **Unlimited Storage:** Scales automatically.
* **High Availability & Durability:** Designed to withstand failures.
* **Secure:** Supports encryption, IAM permissions, bucket policies.
* **Global:** Available in many AWS regions.
* **Cost‑Effective:** Pay only for what you use.

---

## **2. What is an S3 Bucket?**

An **S3 bucket** is a container where you store your objects (files).
Think of it as:

* A **folder** on cloud storage
* Assigned to a **specific AWS region**
* Globally unique bucket name (no other account can use that name)

### **Bucket Contains:**

* Objects (data files)
* Folders (logical grouping, not actual file system)
* Metadata
* Permissions
* Security rules
* Encryption settings
* Versioning info

### **Why do we use S3 buckets?**

* Hosting static websites
* Storing backups, logs, media assets
* Big data analytics storage
* Machine learning datasets
* Application file storage
* Disaster recovery
* CDN origin (CloudFront)

---

## **3. Important S3 Concepts**

### **3.1 Object**

* Actual data stored (images, PDFs, videos, etc.)
* Has metadata and a unique object key

### **3.2 Object Key**

* The complete path to the object inside the bucket

### **3.3 Region**

* S3 bucket must be created in a specific region (ex: Stockholm eu-north‑1)

### **3.4 Access Control**

* Determines who can read/write the bucket
* IAM Policies, Bucket Policies, ACLs

### **3.5 Versioning**

* Keeps multiple versions of the same object
* Helps restore accidentally deleted/overwritten files

### **3.6 Encryption**

* SSE-S3: AWS manages keys
* SSE-KMS: Customer-managed keys

# **4. How to Create an S3 Bucket (Step-by-Step)**

Below is a deep dive into how to set up an S3 bucket—covering all options.

---

## **Step 1: Open AWS Console**

* Sign in to AWS



* Go to **S3** by typing "S3" in the search bar



---

## **Step 2: Create Bucket**

Click **Create bucket**.



### **Required Fields:**

#### **Bucket Name**

* Must be unique globally
* Lowercase only
* Example: *myfirstbucket-sithum-demo*



#### **AWS Region**

Choose a region close to your users.

---

## **Step 3: Bucket Settings Explained**

### **A. Object Ownership**



Controls who owns uploaded objects.

* **ACLs disabled (recommended)** → Bucket owner owns all objects
* **ACLs enabled** → Fine-grained access, but rarely needed

### **B. Block Public Access**

Controls if the bucket or objects can be seen by the public.

* By default, **all public access is blocked** (recommended for security)
* Disable it only if you are hosting a public website

---

## **Step 4: Bucket Features**




### **A. Versioning**

* If enabled → every change creates a new version
* Helps restore deleted/overwritten objects

### **B. Tags** (optional)

* Key-value pairs for organization & billing
* Example: `Project: ML-Data` or `Environment: Production`

### **C. Default Encryption**

Choose how your stored objects are encrypted:

* **SSE-S3**: Simple, AWS-managed keys
* **SSE-KMS**: Customer-controlled keys

### **D. Bucket Key (optional)**

* Reduces KMS API calls
* Useful when encrypting many objects

---

## **Step 5: Advanced Settings**

### **Object Lock** (optional)

Prevent modification/deletion for a set period.

* Must be turned on **during bucket creation** (cannot be enabled later)
* Used for compliance workloads (finance, auditing)

---

## **Step 6: Finalize**


Click **Create Bucket**.

Congratulations! Your bucket is ready.



---

# **6. Additional Deep Dive Topics**

## **6.1 Bucket Policies**

JSON documents that define access permissions.

Example: make bucket public

```json
{
  "Version": "2012-10-17",
  "Statement": [
    {
      "Effect": "Allow",
      "Principal": "*",
      "Action": "s3:GetObject",
      "Resource": "arn:aws:s3:::mybucket/*"
    }
  ]
}
```

---

## **6.2 Lifecycle Rules**

Automate storage management.
Example rules:

* Move objects to **S3 Standard → S3 IA → Glacier** as they age
* Delete old versions
* Clean up expired logs

---

## **6.3 S3 Storage Classes**

Different pricing/performance options:

* **S3 Standard** – frequent access
* **S3 Intelligent-Tiering** – automatic cost optimization
* **S3 Standard-IA** – infrequent access
* **S3 One Zone-IA** – cheaper but single AZ
* **S3 Glacier Instant Retrieval** – archival with fast retrieval
* **S3 Glacier Deep Archive** – extremely low cost, long retrieval time

---

## **6.4 Access Control Methods**

* **IAM Policies** – user/role level
* **Bucket Policies** – bucket level
* **ACLs (not recommended)** – legacy access control

---

## **6.5 S3 Security Best Practices**

* Keep **Block Public Access** ON unless necessary
* Use **SSE-KMS** for sensitive data
* Use **least privilege** IAM policies
* Enable **MFA Delete** for high-security buckets
